```python
def reverse_lookup(d, v):
    for k in d:
        if d[k] == v:
            return k
    raise LookupError('value does not appear in the dictionary')

```

```python
def invert_dict(d):
    inverse = dict()
    for key in d:
     val = d[key]
     if val not in inverse:
        inverse[val] = [key]
    else:
        inverse[val].append(key)
    return inverse
```

```python
hist = histogram('parrotcarrot')
hist
```

```python
inverse = invert_dict(hist)
inverse
```

Python Intermediate — Question 5

Write a function that takes a list of integers and moves all zeros to the end, while maintaining the relative order of the non-zero elements.

```python
nums = [0, 1, 0, 3, 12]
def moving_zero(nums):
    for i in range(len(nums)):
        if nums[i]==0:
            nums.pop(i)
            nums.append(0)
    return(nums)

moving_zero(nums)         

#pop(i) from the middle of a list can take O(n), so your approach can become O(n²).  
    
```

```python
def moving_zero(nums):
    position = 0

    for num in nums:
        if num != 0:
            nums[position] = num
            position += 1

    while position < len(nums):
        nums[position] = 0
        position += 1

    return nums

#time O(n) space O(1)
```

Python Intermediate — Question 7

Write a function that takes a list of integers and returns the longest consecutive sequence length.

```python
nums = [100, 4, 200, 1, 3, 2]
def longest_consecutive(nums):
    num_set = set(nums)
    longest = 0

    for num in num_set:
        if num - 1 not in num_set:
            current = num
            length = 1

            while current + 1 in num_set:
                current += 1
                length += 1

            longest = max(longest, length)

    return longest

longest_consecutive(nums)
```

Python Intermediate — Question 8

Write a Python function that takes a list of integers and returns the most frequent element.

```python
nums = [1, 3, 2, 1, 4, 1, 3, 2]
def most_freq(nums):
    count={}
    for num in nums:
        count[num]=count.get(num,0)+1
    max_key = max(count, key=count.get)
    
    return max_key
        
```

```python
most_freq(nums)
```
