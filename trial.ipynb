```python
def find(word, letter):
    index = 0
    while index < len(word):
        if word[index] == letter:
            return index
        index = index + 1
    return -1
```

```python
def is_abecedarian(word):
    previous = word[0]
    for c in word:
        if c < previous:
            return False
        previous = c
    return True
is_abecedarian(word="Apple")

```

```python
def histogram(s):
    d = dict()
    for c in s:
        if c not in d:
            d[c] = 1
        else:
            d[c] += 1
    return d

h=histogram('summer')
h
```

```python
def reverse_lookup(d, v):
    for k in d:
        if d[k] == v:
            return k
    raise LookupError('value does not appear in the dictionary')

```

```python
h = histogram('parrot')
key = reverse_lookup(h,3)
key
```

```python
def invert_dict(d):
    inverse = dict()
    for key in d:
     val = d[key]
     if val not in inverse:
        inverse[val] = [key]
    else:
        inverse[val].append(key)
    return inverse
```

```python
hist = histogram('parrotcarrot')
hist
```

```python
inverse = invert_dict(hist)
inverse
```

```python
def find_center(rect):
    p = Point()
    p.x = rect.corner.x + rect.width/2
    p.y = rect.corner.y + rect.height/2
    return p
```
