#Hashing

```python
#The returns all numbers that appear more than once, without returning the same duplicate more than once

def find_duplicates(nums):
    container={}
    for num in nums:
        container[num]=container.get(num,0)+1
    print(container)    
    return [num for num, count in container.items() if count > 1]
                

        
    
```

```python
nums = [4, 5, 2, 4, 7, 5, 8, 2, 9, 7]
print(find_duplicates(nums))
```
