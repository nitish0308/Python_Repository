Function returns the unique elements common to both lists.

```python

def common_unique(list1, list2):
    result = [] 
    for num1 in list1:
        for num2 in list2:
            if num1==num2:
                if num1 not in result:
                    result.append(num1)
    return result

#The function you provided currently prints the matching numbers, but it suffers from a few issues: it prints duplicates if a number appears multiple times, it does not actually return the values, and its nested loops make it slow for large datasets (O(N* M) time complexity).
```

```python
list1 = [1, 2, 3, 4, 5, 6]
list2 = [4, 5, 6, 7, 8, 9]
print(common_unique(list1, list2))
```

```python
def common_unique(list1, list2):
    # set() removes duplicates instantly
    # The & operator finds the intersection between sets
    return list(set(list1) & set(list2))

common_unique(list1, list2)
```
