Problem 6. You are given a list of employee records where each record contains (name, department, salary). Write a Python function that returns:

1.The total salary for each department.
2.A set containing employees who appear more than once.

```python
employees = [
    ("Amit", "IT", 70000),
    ("Rahul", "HR", 60000),
    ("Amit", "IT", 80000),
    ("Priya", "IT", 90000),
    ("Rahul", "HR", 65000),
    ("Neha", "Finance", 75000),]
```

```python
def department_summary(employees):
    dept_salary={}
    emp_dept={} 
    duplicate_employees = set()     
    for name, dept, salary in employees:
        dept_salary[dept]=dept_salary.get(dept,0)+salary
        emp_dept[name]=emp_dept.get(name,0)+1
        if emp_dept[name]>1:
            duplicate_employees.add(name)
    
    return dept_salary,duplicate_employees
```

```python
department_summary(employees)
```

Question:
Given a list of numbers, find the first number that appears more than once, and return both:

1. The repeated number
2. The set of all numbers that are repeated

```python
def repeated_num(nums):
    seen = set()
    repeated = set()
    first_repeat = None

    for num in nums:
        if num in seen:
            repeated.add(num)

            if first_repeat is None:
                first_repeat = num
        else:
            seen.add(num)

    return first_repeat, repeated
```

```python
nums = [5, 3, 7, 2, 3, 8, 5, 7, 9]
repeated_num(nums)
```

```python
#2nd approach, with dictonary
def repeated_num(nums):
    frequency = {}
    repeated = set()
    first_repeat = None

    for num in nums:
        frequency[num] = frequency.get(num, 0) + 1

        if frequency[num] == 2:
            repeated.add(num)

            if first_repeat is None:
                first_repeat = num

    return first_repeat, repeated
```

Question:

Given the following transaction records:

transactions = [
    ("A", "Food", 500),
    ("B", "Travel", 1200),
    ("A", "Food", 300),
    ("C", "Food", 700),
    ("B", "Food", 800),
    ("A", "Travel", 1000),
    ("C", "Travel", 400),
    ("B", "Travel", 500),
]
that returns:

1. Total spending by each customer
2. The category in which each customer spent the most
3. A set of customers whose total spending is greater than 1500

```python
transactions = [
    ("A", "Food", 500),
    ("B", "Travel", 1200),
    ("A", "Food", 300),
    ("C", "Food", 700),
    ("B", "Food", 800),
    ("A", "Travel", 1000),
    ("C", "Travel", 400),
    ("B", "Travel", 500),
]
```

```python
def customer_summary(transactions):
    customer_spend = {}
    ex_type_spend = {}

    # Build customer totals and customer-category totals
    for customer, ex_type, amount in transactions:

        # Total spending by customer
        customer_spend[customer] = customer_spend.get(customer, 0) + amount

        # Create inner dictionary for customer
        if customer not in ex_type_spend:
            ex_type_spend[customer] = {}

        # Spending by category for each customer
        ex_type_spend[customer][ex_type] = (
            ex_type_spend[customer].get(ex_type, 0) + amount
        )

    # Find top category for each customer
    top_category = {}

    for customer, categories in ex_type_spend.items():
        top_category[customer] = max(
            categories,
            key=categories.get
        )

    # Find high spenders
    high_spenders = {
        customer
        for customer, total in customer_spend.items()
        if total > 1500
    }

    return {
        "totals": customer_spend,
        "top_category": top_category,
        "high_spenders": high_spenders
    }
```

```python
print(customer_summary(transactions))
```
